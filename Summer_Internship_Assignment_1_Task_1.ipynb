```python
list1 = [1,2,3,4,5,67,8,9,10,1,2,3]
list1
```

```python
list2 = list(set(list1))
list2
```

```python
list3 = sorted(list1)
```

```python
from collections import Counter
```

```python
frequency = dict(Counter(list3))
dict = {
    "min" : min(list3),
    "max" : max(list3),
    "avg" : sum(list3)/len(list3),
    "frequency" : frequency
}
dict
```

```python
print("--- LIST STATISTICS ---")
print(f"Maximum Value: {dict['max']}")
print(f"Minimum Value: {dict['min']}")
print(f"Average Value: {dict['avg']:.2f}")
print("\nElement Frequencies:")
for element, count in dict['frequency'].items():
    print(f"  Element  {element} : Appears {count} time(s)")
```
