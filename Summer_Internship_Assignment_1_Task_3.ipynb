```python
def safe_divide(a, b):
    """
    Divides two numbers and handles errors for division by zero or invalid inputs.

    Args:
        a: The numerator.
        b: The denominator.

    Returns:
        The result of the division or an error message string.
    """
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Error: Cannot divide by zero."
    except TypeError:
        return "Error: Invalid input. Both arguments must be numbers."
    finally:
        print(f"\nDivision executed")

# Test cases
print(f"10 / 2 = {safe_divide(10, 2)}")
print(f"10 / 0 = {safe_divide(10, 0)}")
print(f"10 / 's' = {safe_divide(10, 's')}")
```
