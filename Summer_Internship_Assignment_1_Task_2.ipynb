# File Handling
This notebook demonstrates reading log entries, counting occurrences of each log type, and exporting the results to a JSON file.

```python
import json
from collections import Counter

# 1. Generate or define the log data
log_content = """INFO User logged in
ERROR Database timeout
WARNING Memory usage high
ERROR Invalid credentials
INFO User logged out
ERROR Connection lost"""

with open('logs.txt', 'w') as f:
    f.write(log_content)

print("Sample log file 'logs.txt' created.")
```

```python
# 2. Read the file and count occurrences
log_types = []

with open('logs.txt', 'r') as f:
    for line in f:
        if line.strip():
            # Extract the first word (log level)
            log_type = line.split()[0]
            log_types.append(log_type)

counts = Counter(log_types)
print("Log Counts:", dict(counts))
```

```python
# 3. Save results to a JSON file
with open('log_counts.json', 'w') as jf:
    json.dump(counts, jf, indent=4)

print("Results saved to 'log_counts.json'.")
```

```python
# 4. Print the most frequent log type
if counts:
    most_frequent = counts.most_common(1)[0]
    print(f"The most frequent log type is: {most_frequent[0]} (count: {most_frequent[1]})")
else:
    print("No logs found.")
```
